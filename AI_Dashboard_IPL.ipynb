```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv("/matches.csv")
df.head()
```

```python
df.columns
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
df.isnull().sum()
```

```python
total_matches = len(df)
total_seasons = df['season'].nunique()
total_teams = pd.concat([df['team1'], df['team2']]).nunique()

print("Total Matches:", total_matches)
print("Total Seasons:", total_seasons)
print("Total Teams:", total_teams)
```

```python
matches_per_season = df.groupby('season').size().reset_index(name='Matches')

fig = px.bar(
    matches_per_season,
    x='season',
    y='Matches',
    title='Matches Per Season'
)

fig.show()
```

```python
wins = df['winner'].value_counts().reset_index()

wins.columns = ['team', 'Wins']

fig = px.bar(
    wins,
    x='team',
    y='Wins',
    title='Team Wins'
)

fig.show()
```

```python
toss = df['toss_winner'].value_counts().reset_index()

toss.columns = ['Team', 'Toss Wins']

fig = px.pie(
    toss,
    names='Team',
    values='Toss Wins',
    title='Toss Winners'
)

fig.show()
```

```python
venue = df['venue'].value_counts().head(10).reset_index()

venue.columns = ['Venue', 'Matches']

fig = px.bar(
    venue,
    x='Venue',
    y='Matches',
    title='Top 10 Venues'
)

fig.show()
```

```python
pom = df['player_of_match'].value_counts().head(10).reset_index()

pom.columns = ['Player', 'Awards']

fig = px.bar(
    pom,
    x='Player',
    y='Awards',
    title='Top Player of Match Winners'
)

fig.show()
```

Insights:

1. Mumbai Indians are among the most successful teams.
2. Number of IPL matches increased over seasons.
3. Some venues host significantly more matches.
4. Toss wins do not always translate into match wins.
5. A few players dominate Player of the Match awards.
